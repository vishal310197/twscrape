```python
import asyncio

import pandas as pd
from twscrape import API, gather

KEYWORD = "keittokehotus"
LIMIT = 100
```

```python
async def fetch_tweets(keyword: str, limit: int = 100):
    """Fetch tweets using the accounts already configured in accounts.db."""
    api = API()
    return await gather(api.search(keyword, limit=limit))

tweets = asyncio.run(fetch_tweets(KEYWORD, LIMIT))
len(tweets)
```

```python
rows = [
    {
        "id": tweet.id,
        "username": tweet.user.username,
        "created_at": tweet.date,
        "content": tweet.rawContent,
        "url": tweet.url,
    }
    for tweet in tweets
]

df = pd.DataFrame(rows)
df.head()
```

```python
output_path = "keittokehotus_tweets.csv"
df.to_csv(output_path, index=False)
output_path
```
